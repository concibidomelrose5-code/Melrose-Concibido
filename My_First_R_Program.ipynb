# 5. Numerical Operators

```r
x <- 5
y <- 3

sum <- x + y
sub <- x - y
mul <- x * y
div <- x / y
exp <- x ^ y


print(sum)
print(sub)
print(mul)
print(div)
print(exp)

cat("+", sum)
cat("-", sub)
cat("*", mul)
cat("/", div)
```

# 10. Vectors

```r
grades <- c(90,88,96,83,91)

grades

cat("MEAN:",mean(grades))
cat("\nMAX:",max(grades))
cat("\nMIN:",min(grades))
cat("\nSUM:",sum(grades))
cat("\nLENGTH:",length(grades))
```
